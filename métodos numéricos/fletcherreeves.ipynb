```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import os
import scipy as sp
```

```python
def funcao(x):
    return x[0]**2 + x[1]**2 + x[0]*x[1] + 4*x[0] + 5*x[1]
```

```python
def fletcher_reeves(func, grad, x0, tol=1e-6, max_iter=1000):
    x = x0
    g = grad(x)
    d = -g
    k = 0

    for k in range(max_iter):
        alpha = sp.optimize.line_search(func, grad, x, d)[0]
        if alpha is None:
            alpha = 1e-4  # fallback step size
        x_new = x + alpha * d
        g_new = grad(x_new)
        if np.linalg.norm(g_new) < tol:
            break
        beta = np.dot(g_new, g_new) / np.dot(g, g)
        d = -g_new + beta * d
        x = x_new
        g = g_new


    return x, func(x), k+1
```

```python
ponto_inicial = np.array(input("Digite o ponto inicial separado por vírgula (ex: 0,0): ").split(","), dtype=float)
resultado, valor_funcao, iteracoes = fletcher_reeves(funcao, lambda x: np.array([2*x[0] + x[1] + 4, 2*x[1] + x[0] + 5]), ponto_inicial)
print(f"Resultado: {resultado}")
print(f"Valor da função no resultado: {valor_funcao}")
print(f"Número de iterações: {iteracoes}")
```
